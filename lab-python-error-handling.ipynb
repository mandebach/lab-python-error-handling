# Step 1: Define the function for initializing the inventory with error handling
def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_quantity = False
        while not valid_quantity:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
                if quantity < 0:
                    raise ValueError("Invalid quantity! Please enter a non-negative value.")
                valid_quantity = True
            except ValueError as error:
                print(f"Error: {error}")
        inventory[product] = quantity
    return inventory

````python
def get_customer_orders():
    while True:
        try:
            customer_order_no = int(input("Enter the number of customer orders: "))

            if customer_order_no >= 0:
                list_of_items = [
                    input("Enter the name of a product that a customer wants to order: ")
                    for i in range(customer_order_no)
                ]
                return list_of_items

            else:
                print("Number of orders cannot be negative.")

        except ValueError:
            print("Invalid input. Please enter a valid number.")

list_of_items = get_customer_orders()
````

````python
def total_price(list_of_items):
    total = 0
    for item in list_of_items:
        while True:
            try:
                price = int(input(f"Enter the price for the {item}: "))
            
                if price >= 0:
                    total += price
                    break # You need this to move on to the next item
                else:
                    print("Price cannot be negative. Please enter a valid price.")
            except ValueError:
                print("Invalid input. Please enter a valid number.")
    
    print("The total price is ", total)
    return total

total_price(list_of_items)
````
